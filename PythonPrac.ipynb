```python
print(df[["Name","Department","Salary"]])
```

```python
df.insert=(3,"Bonus", df["Salary"] *0.10)
print(df)
```

```python
##4. Add an Annual_Salary column.
df.insert(7,"Annual_Salary",df["Salary"]*12)
print(df)
```

```python
##5. Remove the Performance column.
df.drop("Performance",axis=1,inplace=True)
print(df)
```

```python
## Display employees who work in the IT department.
emp=df[df["Department"]=="IT"]
print(emp)
```

```python
##Display employees whose salary is greater than ₹40,000.
emp=df[df["Salary"]>40000]
print(emp)
```

```python
##Display IT employees whose salary is greater than ₹35,000.
emp = df[(df["Department"] == "IT") & (df["Salary"] > 35000)]
print(emp)
```

```python
##Display the Name, Department, and Salary of employees whose salary is greater than ₹45,000.
emp = df.loc[df["Salary"] > 45000, ["Name", "Department", "Salary"]]
print(emp)
```

```python
##Display employees who live in Delhi or Mumbai.
emp = df[(df["City"]== "Delhi") | (df["City"]== "Mumbai")]
print(emp)
```

```python
##Sort employees by salary from highest to lowest.
emp = df.sort_values(by="Salary",ascending=False)
print(emp)
```

```python
##Sort employees by experience from lowest to highest.
emp = df.sort_values(by="Experience",ascending=True)
print(emp)
```

```python
## Count the employees in each department.
emp = df.groupby("Department")["Emp_ID"].count()
print(emp)
```

```python
##Calculate the average salary for each department.
emp = df.groupby("Department")["Salary"].mean()
print(emp)
```

```python
##Find the highest salary in each department.
emp = df.groupby("Department")["Salary"].max()
print(emp)
```
